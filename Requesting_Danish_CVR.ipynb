## Forespørgsel til CVR data på datafordeleren.

Oprindeligt bygget interaktivt med [CVR GraphQL](https://datafordeler.dk/dataoversigt/?emne=virksomheder). Where clause med `CVRNummerListe` indbygget programmatisk efterfølgende. Henter data for alle virksomheder i `CVRNummerListe`

```python
query = f"""
       query {{
          CVR_Virksomhed(where: {{ CVRNummer: {{ in: {CVRNummerListe} }} }}) {{
            pageInfo {{
              endCursor
              hasNextPage
            }}
            nodes {{
              CVRNummer
              datafordelerOpdateringstid
              datafordelerRegisterImportSequenceNumber
              datafordelerRowId
              datafordelerRowVersion
              feltliste
              id
              registreringFra
              registreringsaktoer
              registreringTil
              status
              virkningFra
              virkningsaktoer
              virkningTil
              virksomhedOphoersdato
              virksomhedStartdato
            }}
          }}
        }}
        """
```

## Send forespørgsel til Datafordeleren

Sendes som http request:

```python
response = requests.post(url=URL, json={"query": query})
response
```

Udtræk af JSON-data:

```python
result=response.json()
result
```
